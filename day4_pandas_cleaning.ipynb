**Import libraries**

```python
import pandas as pd
```

**Load Csv file used in previous Question**

```python
df = pd.read_csv("/content/tested-selected-columns.csv")
df
```

**identify missing values**

```python
df.isnull().sum()
```

**Handle missing values**

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Fare"] = df["Fare"].fillna(df["Fare"].median())
```

```python
df.isnull().sum()
```

**checking datatypes**

```python
df.dtypes
```

**Removing duplicates**

```python
df = df.drop_duplicates()
```

**Adding Dataframe**

```python
class_info = pd.DataFrame({
    "Pclass": [1, 2, 3],
    "Class_Label": ["Premium", "Standard", "Economy"]
})

class_info
```

**merging with dataset**

```python
df.merge(class_info)
```
