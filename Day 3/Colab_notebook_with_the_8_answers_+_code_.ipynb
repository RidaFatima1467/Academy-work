```python
import pandas as pd
df= pd.read_csv("/content/Amazon.csv")
df
```

**1. Total Sales of company?**

```python
total_sales = df['TotalAmount'].sum()

print("Total Sales:", total_sales)
```

**2. Which category of product made high sales?**

```python
category_sales = df.groupby('Category')['TotalAmount'].sum().sort_values(ascending=False)
category_sales
```

**3. Checking Which Country generate highest sales?**

```python
country_sales = df.groupby('Country')['TotalAmount'].sum().sort_values(ascending=False)
country_sales
```

**4. Commonly Used payment method in customers?**

```python
payment_method = df['PaymentMethod'].value_counts()
payment_method
```

**5. Find the average order value?**

```python
average_order = df['TotalAmount'].mean().astype(int)
average_order
```

**6. Which is generate highest sale in company?**

```python
top_product = df.groupby('ProductName')['TotalAmount'].sum().sort_values(ascending=False).head(10)
top_product
```

**7. Which Brand make highest sales?**

```python
Brand_name = df.groupby('Brand')['TotalAmount'].sum().sort_values(ascending=False).head(10)
Brand_name
```

**8. How much order returned or canceled?**

```python
returned_orders = (df['OrderStatus'] == 'Returned').sum()

cancelled_orders = (df['OrderStatus'] == 'Cancelled').sum()

print(returned_orders)

print(cancelled_orders)
```
