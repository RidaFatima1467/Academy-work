**Cleaning df1 dataset**

**Making copy of df1 dataset**

```python
df1_clean = df1.copy()
```

**Removing text spaces from text columns**

```python
for col in df1_clean.select_dtypes(include="object").columns:
    df1_clean[col] = df1_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

**Count rows before cleaning**

```python
df1_before = len(df1_clean)

print("Rows before cleaning:", df1_before)
```

**Count duplicate rows and remove them**

```python
duplicate_count = df1_clean.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
df1_clean = df1_clean.drop_duplicates()
print("Duplicate rows removed:", duplicate_count)
```

**Rows after cleaning**

```python
print("Rows after cleaning:", len(df1_clean))
```

**Covert ID into Numeric**

```python
df1_clean["id"] = pd.to_numeric(
    df1_clean["id"],
    errors="coerce"
)
```

**convert into int**

```python
df1_clean["id"] = df1_clean["id"].astype("Int64")
```

**Convert date columns into datetime dtype**

```python
df1_clean["born_date"] = pd.to_datetime(
    df1_clean["born_date"],
    errors="coerce"
)

df1_clean["debut_date"] = pd.to_datetime(
    df1_clean["debut_date"],
    errors="coerce"
)

df1_clean["last_date"] = pd.to_datetime(
    df1_clean["last_date"],
    errors="coerce"
)
```

**check ivalid weight count**

```python
invalid_weight_count = (df1_clean["weight"] <= 0).sum()

print("Invalid weights:", invalid_weight_count)
```

```python
df1_clean.loc[df1_clean["weight"] <= 0, "weight"] = np.nan
print("Invalid weights fixed:", invalid_weight_count)
```

**Finding mediam weight and fill with median**

```python
weight_median = df1_clean["weight"].median()

print("Median weight:", weight_median)

df1_clean["weight"] = df1_clean["weight"].fillna(weight_median)
```

**Handeling missing profile pic**

```python
df1_clean["profile_pic"] = df1_clean["profile_pic"].fillna("Not available")
```

**Handeling missing common names**

```python
df1_clean["player_common_names"] = df1_clean["player_common_names"].fillna("None")
```

**Handeling missing team info**

```python
df1_clean["player_teams"] = df1_clean["player_teams"].fillna("Not available")
```

**DF1 after handling missing values now 0 null values**

```python
print("Players Info — missing values after cleaning:")

display(df1_clean.isna().sum().sort_values(ascending=False).to_frame("missing"))
```

**Checking again datatypes**

```python
print("Players Info — final data types:")

display(df1_clean.dtypes.to_frame("dtype"))
```

**Save cleaned Players dataset**

```python
df1_clean.to_csv("cleaned_players_info.csv", index=False)
```

**Cleaning df2 dataset**

**Making copy of df2 dataset**

```python
df2_clean = df2.copy()
```

**Rows before cleaning**

```python
df2_before = len(df2_clean)

print("Rows before cleaning:", df2_before)
```

**Check duplicate rows**

```python
df2_duplicates = df2_clean.duplicated().sum()

print("Duplicate rows:", df2_duplicates)
```

**Remove extra spaces from text columns**

```python
for col in df2_clean.select_dtypes(include="object").columns:
    df2_clean[col] = df2_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

**Clean player_id**

```python
df2_clean["player_id"] = (
    df2_clean["player_id"]
    .astype(str)
    .str.extract(r"(\d+)")[0]
)
```

**Convert player_id to integer**

```python
df2_clean["player_id"] = pd.to_numeric(
    df2_clean["player_id"],
    errors="coerce"
).astype("Int64")
```

**Convert year to integer**

```python
df2_clean["year"] = pd.to_numeric(
    df2_clean["year"],
    errors="coerce"
).astype("Int64")
```

**Remove duplicate rows**

```python
df2_clean = df2_clean.drop_duplicates().copy()
```

**Check invalid negative games**

```python
invalid_games = (
    df2_clean["games_played"] < 0
).sum()

print("Invalid negative games:", invalid_games)
```

**Remove negative games**

```python
df2_clean = df2_clean[
    df2_clean["games_played"] >= 0
].copy()
```

**Check negative fantasy points**

```python
print(df2_clean["total_fantasy_points"].dtype)
```

```python
print("Negative values:")
display(
    df2_clean[df2_clean["total_fantasy_points"] < 0][
        ["player_id", "year", "games_played", "avg_fantasy_points", "total_fantasy_points"]
    ]
)
```

```python
df2_clean.loc[
    df2_clean["total_fantasy_points"] < 0,
    "total_fantasy_points"
] = 0
```

**Replace negative fantasy points with missing value**

```python
df2_clean.loc[
    df2_clean["avg_fantasy_points"] < 0,
    "avg_fantasy_points"
] = 0
```

```python
df2_clean["total_fantasy_points"] = df2_clean[
    "total_fantasy_points"
].fillna(
    df2_clean["avg_fantasy_points"] * df2_clean["games_played"]
)
```

**List the total/count columns & Convert these columns to numeric**

```python
count_total_cols = [
    "kicks",
    "marks",
    "handballs",
    "disposals",
    "goals",
    "behinds",
    "hit_outs",
    "tackles",
    "rebound_50s",
    "inside_50s",
    "clearances",
    "clangers",
    "free_kicks_for",
    "free_kicks_against",
    "brownlow_votes",
    "contested_possessions",
    "uncontested_possessions",
    "contested_marks",
    "marks_inside_50",
    "one_percenters",
    "bounces",
    "goal_assists",
    "total_score"
]
for col in count_total_cols:
    df2_clean[col] = pd.to_numeric(
        df2_clean[col],
        errors="coerce"
    )
```

**Fill missing values with 0**

These are count/total statistics, so missing values are treated as zero according to our cleaning rule.

```python
for col in count_total_cols:
    df2_clean[col] = df2_clean[col].fillna(0)
```

**Convert total percentage played to numeric**

```python
df2_clean["total_percentage_played"] = pd.to_numeric(
    df2_clean["total_percentage_played"],
    errors="coerce"
)
```

**Fill missing percentage played and remaining with 0**

```python
df2_clean["total_percentage_played"] = (
    df2_clean["total_percentage_played"]
    .fillna(
        df2_clean["avg_percentage_played"] *
        df2_clean["games_played"]
    )
)
```

```python
df2_clean["total_percentage_played"] = (
    df2_clean["total_percentage_played"]
    .fillna(0)
)
```

**Define average columns**

```python
avg_to_total = {
    "avg_kicks": "kicks",
    "avg_marks": "marks",
    "avg_handballs": "handballs",
    "avg_disposals": "disposals",
    "avg_goals": "goals",
    "avg_behinds": "behinds",
    "avg_hit_outs": "hit_outs",
    "avg_tackles": "tackles",
    "avg_rebound_50s": "rebound_50s",
    "avg_inside_50s": "inside_50s",
    "avg_clearances": "clearances",
    "avg_clangers": "clangers",
    "avg_free_kicks_for": "free_kicks_for",
    "avg_free_kicks_against": "free_kicks_against",
    "avg_contested_possessions": "contested_possessions",
    "avg_uncontested_possessions": "uncontested_possessions",
    "avg_contested_marks": "contested_marks",
    "avg_marks_inside_50": "marks_inside_50",
    "avg_one_percenters": "one_percenters",
    "avg_bounces": "bounces",
    "avg_goal_assists": "goal_assists",
    "avg_score": "total_score",
    "avg_fantasy_points": "total_fantasy_points",
    "avg_percentage_played": "total_percentage_played"
}
```

**Recalculate averages**

```python
for avg_col, total_col in avg_to_total.items():

    df2_clean[avg_col] = np.where(
        df2_clean["games_played"] > 0,
        df2_clean[total_col] /
        df2_clean["games_played"],
        0
    ).round(1)
```

**Standardize team names**

```python
df2_clean["team"] = (
    df2_clean["team"]
    .astype(str)
    .str.strip()
    .str.title()
)
```

**Check missing values after cleaning**

```python
print("Missing values after cleaning:")

display(
    df2_clean.isna()
    .sum()
    .sort_values(ascending=False)
    .to_frame("missing")
)
```

**Check final data types**

```python
print("Final data types:")

display(
    df2_clean.dtypes.to_frame("dtype")
)
```

**Merging**

**Load clean datasets**

```python
import pandas as pd

df1_clean = pd.read_csv("/content/cleaned_players_info.csv")
df2_clean = pd.read_csv("/content/cleaned_seasonal_stats.csv")

print("Players Info:", df1_clean.shape)
print("Seasonal Stats:", df2_clean.shape)
```

```python
df1_clean["id"] = pd.to_numeric(df1_clean["id"], errors="coerce")
df2_clean["player_id"] = pd.to_numeric(df2_clean["player_id"], errors="coerce")
```

```python
player_ids = set(df1_clean["id"].dropna())
stats_ids = set(df2_clean["player_id"].dropna())

unmatched_ids = sorted(stats_ids - player_ids)

print("Unique IDs in Players Info:", len(player_ids))
print("Unique IDs in Seasonal Stats:", len(stats_ids))
print("Unmatched player IDs:", len(unmatched_ids))

print("\nFirst 50 unmatched IDs:")
print(unmatched_ids[:50])
```

```python
merged_players = df2_clean.merge(
    df1_clean,
    left_on="player_id",
    right_on="id",
    how="inner"
)

merged_players = merged_players.drop_duplicates()

merged_players.to_csv(
    "/content/merged_players.csv",
    index=False
)

print("Final merged CSV saved!")
```

**Check duplicates after merging**
