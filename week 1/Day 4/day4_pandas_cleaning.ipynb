**What is the difference between .drop() and .dropna()?**

.drop()is used to delete the specific row and column.

.dropna() is used to drop the row and column which contain null values.

**How do you change the dtype of a column — and when would you need to?**

 By using **astype** we can change the dtype of the column. we can change the dtype when the column values are in float and we need them in integer then we need it.
 For example: age must be in integer values. if in age column float values present we can apply astype to change it into integer.

**What does .apply() do and how is it different from vectorized operations?**

.apply() can run a function on values like a loop. whereas the vectorized operation perform function or operation on entire array at once.

.apply() is used for complex operations or calculations also work slowly  whereas the vectorized operations perform simple calculations and is fast.

**Difference between .pivot() and .pivot_table()?**

.pivot() is used to reshape the entire data but it doesn't handle the duplicates and null values.

.pivot_table is also to reshape the data but it can handle the missing values as well using aggregation operations like mean,median etc.

**What does .merge() do and what are the 4 types of joins?**

.merge() is used to combine dataframes or datasets.

1. inner join:
 It is used to join or combine the same data of both the dataframes or datasets.

 2. left join:
  It will combine all the data of the left dataframes and combine the entities which are same in left dataframe and the right dataframe.

  3. right join:
  complete right dataframe or dataset and matching entities in both left and right dataframe.

  4. outer join:
  combine both the datasets or dataframes completely.

**Import libraries**

```python
import pandas as pd
```

**Load amazon csv file**

```python
df = pd.read_csv("/content/Amazon.csv")
df
```

**Identify missing values**

```python
df.isnull().sum()
```

There are no missing values are present in dataset if present then drop the rows if row is not necessary otherwise filled with mean, median,mode,backward fill etc.

**Check datatypes**

```python
df.dtypes
```

**fixing datatypes**

```python
# coverting datetime column
df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="coerce")

#ensuring numeric columns are in numeric form or not
numeric_columns = [
    "Quantity", "UnitPrice", "Discount", "Tax",
    "ShippingCost", "TotalAmount"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nData types after fixing:")
print(df.dtypes)
```

**Removing Duplicates**

```python
df.duplicated().sum()
```

**Creating to derived columns**
1. GrossSales = Quantity*UnitPrice
sale value before discount,tax etc
2. DiscountAmount = GrossSales*Discount
convert acutal discount percentage into actual money

```python
df["GrossSales"] = df["Quantity"] * df["UnitPrice"]
df["DiscountAmount"] = df["GrossSales"] * df["Discount"]

df[[
    "Quantity", "UnitPrice", "Discount",
    "GrossSales", "DiscountAmount"
]]
```

**Reshape data by groupby and aggregation**

Grouping by category helps the business compare the performance of diiferent category of products.

```python
category_summary = (
    df.groupby("Category")
      .agg(
          OrderCount=("OrderID", "count"),
          TotalQuantity=("Quantity", "sum"),
          TotalGrossSales=("GrossSales", "sum"),
          TotalDiscount=("DiscountAmount", "sum"),
          AverageOrderAmount=("TotalAmount", "mean")
      )
      .reset_index()
)

category_summary
```

**Creating dataframe and merging it with dataset**

```python
#new DataFrame
category_info = pd.DataFrame({
    'Category': ['Electronics', 'Clothing', 'Furniture', 'Beauty'],
    'Category_Type': ['Technology', 'Fashion', 'Home', 'Personal Care']
})

# Merge with the dataset
df = df.merge(category_info, on='Category', how='left')
df

```
